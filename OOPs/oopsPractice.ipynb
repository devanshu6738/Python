```python
class Car:
    def __init__(self,color,brand,price):
        self.color=color
        self.brand=brand
        self.price=price

    def display(self):
        print(f"color : {self.color}, brand : {self.brand} , price : {self.price}")

        
```

```python
c1=Car("green","Lamborgini",20000000)
```

```python
c1.display()
```

```python
class Student:
    def __init__(self,name,age,grade):
        self.name=name
        self.age=age
        self.grade=grade

    @classmethod
    def change_school(cls,new_class):
        cls.school_name=new_class
    
    def display(self):
        print(f"name : {self.name}, brand : {self.age} , price : {self.grade}")

class StudentManager:
    def __init__(self):
        self.students=[]

    def add_student(self,name,age,grade):
        student=Student(name,age,grade)
        self.students.append(student)

    def display_all(self):
        for student in self.students:
            student.display()

    
```

```python
s1=Student("Devanshu","21",99)
s1.change_school="carmel"
```

```python
manager=StudentManager()
manager.add_student("Ayush","21",100)
manager.add_student("Dev","21",93)
```

```python
manager.display_all()
```
